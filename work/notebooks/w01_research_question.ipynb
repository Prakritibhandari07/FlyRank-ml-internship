## 2. The Question

**Research question:** Which content pages should be prioritized for
review first, based on signs of decline combined with enough visibility
to matter?

**Unit of analysis:** One page (one `content_id`) — I'm ranking individual
content items, not clients or days.

**Output:** A ranked list of pages, ordered by a review-priority score,
each with a reason code explaining why it was flagged (e.g.,
"stale_visible_page" or "declining_with_demand").

**The action someone could take:** A content reviewer works down the
ranked list and decides whether to refresh, expand, or otherwise update
each flagged page, instead of reviewing pages randomly or by gut feeling.

**Cost of a wrong recommendation:** If the model flags a page that isn't
actually worth reviewing, the reviewer wastes limited time on a low-value
page instead of a genuinely declining one — this is a capacity/opportunity
cost, not a safety-critical error, but it still matters because reviewer
time is limited.

**Why data/ML can help:** A hand-written rule can only combine a few
signals in a fixed way. A learned model can weigh many signals (position,
CTR, freshness, word count, engagement) together and adapt to patterns a
simple rule would miss — which is exactly what the Week 1 comparison
showed (random forest far outperformed the hand-written rule on
precision@50).
